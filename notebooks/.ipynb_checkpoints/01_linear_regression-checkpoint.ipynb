```python
def mse(y_true, y_pred):
    loss_function = 0
    for i in range(len(y_true)):
        loss_function += (y_true[i] - y_pred[i]) ** 2

    return loss_function / len(y_true)
```

```python
def part_derivative(y_true, X, theta, b):
    predictions = X @ theta + b
    error = predictions - y_true
    gradient = X.T @ error
    return (2 / len(y_true)) * gradient
```

```python
def grad_desc(theta, lr, epochs, y_true, X, b):

    losses = []

    for epoch in range(epochs):

        predictions = X @ theta + b
        loss = mse(y_true, predictions)
        losses.append(loss)
        db = (2 / len(y_true)) * np.sum(predictions - y_true)
        gradient = part_derivative(y_true, X, theta, b)
        theta -= lr * gradient
        b -= lr * db

    return theta
```

```python
theta = np.zeros(8)
b = 0

gradient = part_derivative(y_true, X_true, theta, b)
print(gradient)
print(np.max(np.abs(gradient)))
```

```python
epochs = 1000
lr = 0.001
theta = np.zeros(8)
b = 0

params = grad_desc(theta, lr, epochs, y_true, X_true, b)
params
```
